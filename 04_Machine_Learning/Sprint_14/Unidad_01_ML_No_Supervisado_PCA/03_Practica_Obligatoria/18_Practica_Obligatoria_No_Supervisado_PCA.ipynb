Championship". Estas microcámaras tienen la peculiaridad de que son capaces de tomar fotos
encuadradas de las caras y la desventaja de que no tienen un gran ancho de banda de

### #2 MODELO PARA LAS MICROCÁMARAS
**Objetivo:** Construir un modelo que pueda funcionar en las microcámaras, es decir que
pueda funcionar con datos comprimidos.

Para cumplir con el objetivo se os ocurre emplear la doble propiedad de la PCA, que
permite comprimir datos y mantener la capacidad informativa de estos. Sigue los siguientes

3. Para el rango anterior entrena un modelo de clasificación y apunta su scoring en una

```python
X_train.shape
```

```python
pca = PCA()  # int -> nº de PCs; float entre 0 y 1 -> nº de PCs para esa varianza acumulada mínima
pca.fit(X_train)
```

```python
len(X_train)
```

```python
pca.n_components_  # num_features >> num_instancias: alta dimensionalidad, PCA limitada por el nº de instancias
```

Ahora vamos a ver las pcs necesarias para 10 valores entre 0.2% y 2.5% de compresión

```python
# Demostración tangible de compresión: guardamos el dataset completo y una versión
# con solo 8 columnas (~0.2% de las 4096 originales) para comparar tamaños en disco.
X_train.to_csv("prueba.csv")
X_train.iloc[:, 0:8].to_csv("prueba_comprimida.csv")
```

```python
compresion_minima_pct = 0.2
compresion_maxima_pct = 2.5

num_pcs_a_probar = []
num_features_totales = len(X_train.columns)  # 4096, por florear un poco

for porcentaje in np.linspace(compresion_minima_pct, compresion_maxima_pct, 5):
    # grado_compresion = num_pc / num_features * 100 -> num_pc = grado_compresion * num_features / 100
    num_pc = int(porcentaje * num_features_totales / 100)
    print(f"Para {porcentaje:0.3f}% de compresión, voy a probar {num_pc} PCs")
    num_pcs_a_probar.append(num_pc)

print("Rangos que voy a probar:")
print(num_pcs_a_probar)
```

```python
X_train_pca = pca.transform(X_train)  # array de numpy
X_test_pca = pca.transform(X_test)  # con la misma proyección de train, no reajustamos sobre test

X_train_pca.shape
```

```python
X_test_pca.shape
```

```python
num_pcs_a_probar
```

```python
registros_resultado = []

for num_pc in num_pcs_a_probar:
    X_train_recortado = X_train_pca[:, :num_pc]
    X_test_recortado = X_test_pca[:, :num_pc]

    modelo_rf = RandomForestClassifier(random_state=42)
    score_cv = np.mean(cross_val_score(modelo_rf, X_train_recortado, y_train, cv=5, scoring="balanced_accuracy"))

    modelo_rf.fit(X_train_recortado, y_train)
    y_pred = modelo_rf.predict(X_test_recortado)

    score_test = recall_score(y_test, y_pred, average="macro")

    registros_resultado.append({
        "num_pc": num_pc,
        "factor_compresión": round(num_pc / num_features_totales * 100, 3),
        "eval_cv": round(score_cv * 100, 3),
        "baseline_cv": round(baseline_cv * 100, 3),
        "diff_cv": round((baseline_cv - score_cv) * 100, 3),
        "eval_test": round(score_test * 100, 3),
        "baseline_test": round(baseline_test * 100, 3),
        "diff_test": round((baseline_test - score_test) * 100, 3),
    })
```

```python
registros_resultado
```

```python
df_resultados = pd.DataFrame(registros_resultado)
df_resultados
```

```python
plt.figure(figsize=(8, 8))
umbral_estricto = 3
umbral_relajado = 5

for columna in ["diff_cv", "diff_test"]:
    plt.plot(df_resultados["factor_compresión"], df_resultados[columna], label=columna)

plt.xlabel("Factor Compresión")
plt.ylabel("Diff con baseline")
plt.axhline(y=umbral_estricto, color="r", linestyle="-", label=f"Umbral {umbral_estricto} pp")
plt.axhline(y=umbral_relajado, color="b", linestyle="--", label=f"Umbral {umbral_relajado} pp")

plt.legend()
plt.show()
```

```python
# AND, no OR: ambas métricas deben cumplir el umbral según el enunciado
# ("tanto en CV como en test"). Fallback a umbral_relajado si ninguna fila cumple.
candidatas = (df_resultados["diff_cv"] <= umbral_estricto) & (df_resultados["diff_test"] <= umbral_estricto)
if df_resultados[candidatas].empty:
    candidatas = (df_resultados["diff_cv"] <= umbral_relajado) & (df_resultados["diff_test"] <= umbral_relajado)

df_resultados[candidatas]
```

### #3 COMPRESION PARA CLASIFICACION POSTERIOR

**Objetivo**: Obtener el número de componentes que permita una compresión menor y al
tiempo que el modelo en el servidor central no baje su rendimiento respecto a no usar
imágenes comprimidas.

Para esta parte la idea que se os ha ocurrido es emplear también la PCA como
compresor ya que así siempre podrían pasar a la opción anterior si eso fuese suficiente.
Pero en este caso no vamos a utilizar el dataset comprimido con las PCAs para detectar las
caras, sino el dataset una vez descomprimido (recuerda que puede emplear `inverse_transform`
para "descomprimir"). Los pasos a seguir son:

1. Escoge un rango de valores que permitan una compresión aún mayor (recuerda que el ancho
   de banda es mínimo) entre el 1 por mil y el 1 por ciento. Escoge 5 valores de número de
   PCAs que permitan movernos en ese rango.
2. Para cada uno de esos valores: aplica la PCA al X_train, obten un X_train_unzipped
   aplicando la inversa de la PCA y entrena un modelo de clasificación y pruébalo contra
   test, apunta el balanced accuracy.
3. Crea un dataframe o haz un visualización comparando como es la medidad de balance
   accuracy para cada valor de número de pcas escogido y cuál su factor de compresión.
4. Sabiendo que no podemos perder más de 3 puntos porcentuales respecto al baseline, ¿qué
   numero de PCA escogerías?

```python
np.linspace(0.1, 1, 5)
```

```python
compresion_minima_pct = 0.1
compresion_maxima_pct = 1

num_pcs_a_probar = []
num_features_totales = len(X_train.columns)  # 4096, por florear un poco

for porcentaje in np.linspace(compresion_minima_pct, compresion_maxima_pct, 5):
    # grado_compresion = num_pc / num_features * 100 -> num_pc = grado_compresion * num_features / 100
    num_pc = int(porcentaje * num_features_totales / 100)
    num_pcs_a_probar.append(num_pc)

print("Rangos que voy a probar:")
print(num_pcs_a_probar)
```

```python
registros_resultado = []

for num_pc in num_pcs_a_probar:
    pca_temp = PCA(num_pc)  # hay que "comprimir" de nuevo, no podemos reutilizar X_train_pca directamente
    X_comprimido = pca_temp.fit_transform(X_train)
    X_comprimido_test = pca_temp.transform(X_test)

    X_train_descomprimido = pca_temp.inverse_transform(X_comprimido)  # comprimimos y luego "descomprimimos"
    print("num pcas:", num_pc, "num features descomprimidas: ", X_train_descomprimido.shape[1])
    X_test_descomprimido = pca_temp.inverse_transform(X_comprimido_test)  # lo mismo para test

    modelo_rf = RandomForestClassifier(random_state=42)
    score_cv = np.mean(cross_val_score(modelo_rf, X_train_descomprimido, y_train, cv=5, scoring="balanced_accuracy"))

    modelo_rf.fit(X_train_descomprimido, y_train)
    y_pred = modelo_rf.predict(X_test_descomprimido)
    score_test = recall_score(y_test, y_pred, average="macro")

    registros_resultado.append({
        "num_pc": num_pc,
        "factor_compresión": round(num_pc / num_features_totales * 100, 2),
        "eval_cv": round(score_cv * 100, 2),
        "baseline_cv": round(baseline_cv * 100, 2),
        "diff_cv": round((baseline_cv - score_cv) * 100, 2),
        "eval_test": round(score_test * 100, 2),
        "baseline_test": round(baseline_test * 100, 2),
        "diff_test": round((baseline_test - score_test) * 100, 2),
    })
```

```python
df_resultados = pd.DataFrame(registros_resultado)
df_resultados
```

```python
plt.figure(figsize=(8, 8))
umbral_estricto = 3
umbral_relajado = 5

for columna in ["diff_cv", "diff_test"]:
    plt.plot(df_resultados["factor_compresión"], df_resultados[columna], label=columna)

plt.xlabel("Factor Compresión")
plt.ylabel("Diff con baseline")
plt.axhline(y=umbral_estricto, color="r", linestyle="-", label=f"Umbral {umbral_estricto} pp")
plt.axhline(y=umbral_relajado, color="b", linestyle="--", label=f"Umbral {umbral_relajado} pp")

plt.legend()
plt.show()
```

```python
# AND, no OR; mismo umbral (3 pp) para ambas métricas según el enunciado de esta sección.
candidatas = (df_resultados["diff_cv"] <= 3) & (df_resultados["diff_test"] <= 3)
df_resultados[candidatas]
```

```python
def pintar_caras(caras, columnas=5, titulo=""):
    caras = caras.reshape(-1, 64, 64)
    filas = -(-len(caras) // columnas)  # ceil division
    plt.figure(figsize=(columnas * 4, filas * 4.4))
    for indice, cara in enumerate(caras):
        plt.subplot(filas, columnas, indice + 1)
        plt.imshow(cara, cmap="gray")
        plt.axis("off")
    plt.title(titulo)
    plt.show()
```
